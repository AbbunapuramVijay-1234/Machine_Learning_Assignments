```python
import pandas as pd

# 1. Create the DataFrame
d = {
    'player': ['M S Dhoni', 'hardhik', 'rahul', 'Mohit sharma', 'Andrerussel', 'jashprret', 'virat kohli', 'rohit sharma'],
    'Team': ['Chennai Super King', 'Mumbai indians', 'kings eleven', 'GT', 'kolkata knight riders', 'Mumbai indians', 'RCB', 'Mumbai indians'],
    'category': ['Batsman', 'Batsman', 'Batsman', 'Bowler', 'Bowler', 'Batsman', 'Batsman', 'Batsman'],
    'Bidprice': [13, 11, 10, 12, 7, 10, 17, 15],
    'Runs': [4000, 2400, 3000, 1000, 1500, 3500, 5000, 3900]
}

df = pd.DataFrame(d)
```

```python
display(df)
```

# **1.Display All Bowlers**

```python
bowlers=df[df['category']=='Bowler']
print(bowlers)
```

# **2. Display all players of Mumbai Indians**

```python
mumbai_players = df[df['Team'] == 'Mumbai indians']
print(mumbai_players)
```

# **3. Display all Mumbai Indian players with runs greater than 3000**

```python
mumbai_high_runs = df[(df['Team'] == 'Mumbai indians') & (df['Runs'] > 3000)]
print(mumbai_high_runs)
```

# **4. Display the details of players who play for RCB or Chennai Super King**

```python
rcb_csk_players = df[df['Team'].isin(['RCB', 'Chennai Super King'])]
print(rcb_csk_players)
```

# **5. Print the average bid price and run**

```python
print("Average Bid Price:", df['Bidprice'].mean())
print("Average Runs:", df['Runs'].mean())
```

# **6. Sort the player by the bidding price**

```python
sorted_players = df.sort_values(by='Bidprice', ascending=False)
print(sorted_players)
```

# **7. Find most expensive player**

```python
most_expensive = df.loc[df['Bidprice'].idxmax()]
print(most_expensive)
```

# **8. Find the player with lowest run**

```python
lowest_run_player = df.loc[df['Runs'].idxmin()]
print(lowest_run_player)
```

# **9. Print the total player/no. of rows in a dataset**

```python
print("Total number of rows/players:", len(df))
```

# **10. Print total players per each team**

```python
players_per_team = df['Team'].value_counts()
print(players_per_team)
```

# **11. Find player who had highest bid price from each team**

```python
highest_bid_per_team = df.loc[df.groupby('Team')['Bidprice'].idxmax()]
print(highest_bid_per_team)
```

# **12. Find the average run for each team**

```python
avg_run_per_team = df.groupby('Team')['Runs'].mean()
print(avg_run_per_team)
```
